```python

# STEP 2: Import and verify libraries

import pandapower as pp
import pandapower.networks as pn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Pandapower version:", pp.__version__)
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)

print("\nAll libraries imported successfully.")
```

```python

# STEP 3: Load the IEEE 14-bus test system

net = pn.case14()

print("IEEE 14-Bus Network Loaded Successfully")

print("\nNetwork Information:")
print("Number of buses:", len(net.bus))
print("Number of lines:", len(net.line))
print("Number of transformers:", len(net.trafo))
print("Number of generators:", len(net.gen))
print("Number of external grids:", len(net.ext_grid))
print("Number of loads:", len(net.load))
```

```python
# STEP 4: Inspect network data

print("BUS DATA")
display(net.bus)

print("LINE DATA")
display(net.line)

print("GENERATOR DATA")
display(net.gen)

print("LOAD DATA")
display(net.load)
```

```python
# STEP 5: Run baseline power-flow analysis

pp.runpp(
    net,
    algorithm="nr",
    calculate_voltage_angles=True
)

if net.converged:
    print("BASELINE LOAD FLOW CONVERGED SUCCESSFULLY")
else:
    print("BASELINE LOAD FLOW DID NOT CONVERGE")
```

```python
# STEP 6: Extract bus voltage results

bus_results = net.bus[["name"]].join(net.res_bus)

print("BASELINE BUS VOLTAGE PROFILE")

display(
    bus_results[["name", "vm_pu", "va_degree"]]
)
```

```python
# STEP 7: Calculate transmission losses

line_losses = net.res_line["pl_mw"].sum()

transformer_losses = (
    net.res_trafo["pl_mw"].sum()
    if len(net.res_trafo) > 0 else 0
)

total_losses = line_losses + transformer_losses

print("BASELINE SYSTEM LOSSES")
print("Line losses:", round(line_losses, 5), "MW")
print("Transformer losses:", round(transformer_losses, 5), "MW")
print("Total active power losses:", round(total_losses, 5), "MW")
```

```python
# STEP 8: Calculate total load demand

total_active_load = net.load["p_mw"].sum()
total_reactive_load = net.load["q_mvar"].sum()

print("TOTAL LOAD DEMAND")
print("Active power demand:", round(total_active_load, 4), "MW")
print("Reactive power demand:", round(total_reactive_load, 4), "MVAr")
```

```python

# STEP 9: Plot baseline voltage profile

plt.figure(figsize=(10, 5))

bus_numbers = np.arange(1, len(net.bus) + 1)
voltages = net.res_bus["vm_pu"].values

plt.plot(
    bus_numbers,
    voltages,
    marker="o",
    linewidth=2,
    label="Baseline voltage"
)

plt.axhline(1.0, linestyle="--", label="Reference voltage")
plt.axhline(0.95, linestyle=":", label="Lower reference limit")
plt.axhline(1.05, linestyle=":", label="Upper reference limit")

plt.xlabel("Bus Number")
plt.ylabel("Voltage Magnitude (p.u.)")
plt.title("Baseline Voltage Profile – IEEE 14-Bus System")

plt.xticks(bus_numbers)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print("Minimum voltage:", round(voltages.min(), 4), "p.u.")
print("Maximum voltage:", round(voltages.max(), 4), "p.u.")
```

```python

# STEP 11: Generate IEEE 14-Bus Single-Line Diagram

import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Create network graph
G = nx.MultiGraph()

# Add buses
for bus in net.bus.index:
    G.add_node(bus)

# Extract transmission line connections
line_edges = [
    (row.from_bus, row.to_bus)
    for _, row in net.line.iterrows()
]

# Extract transformer connections
trafo_edges = [
    (row.hv_bus, row.lv_bus)
    for _, row in net.trafo.iterrows()
]

G.add_edges_from(line_edges)
G.add_edges_from(trafo_edges)

# Generate automatic network layout
pos = nx.spring_layout(G, seed=14, k=1.3, iterations=200)

# Identify generator, external grid and load buses
gen_buses = set(net.gen.bus)
ext_buses = set(net.ext_grid.bus)
load_buses = set(net.load.bus)

# Draw diagram
plt.figure(figsize=(13, 9))

# Transmission lines
nx.draw_networkx_edges(
    G, pos,
    edgelist=line_edges,
    edge_color="steelblue",
    width=2
)

# Transformers
nx.draw_networkx_edges(
    G, pos,
    edgelist=trafo_edges,
    edge_color="darkorange",
    width=2.5,
    style="dashed"
)

# All buses
nx.draw_networkx_nodes(
    G, pos,
    node_size=650,
    node_color="lightblue",
    edgecolors="black"
)

# Generator buses
nx.draw_networkx_nodes(
    G, pos,
    nodelist=list(gen_buses),
    node_shape="s",
    node_color="tomato",
    node_size=700
)

# External grid bus
nx.draw_networkx_nodes(
    G, pos,
    nodelist=list(ext_buses),
    node_shape="*",
    node_color="gold",
    node_size=1100
)

# Load buses
nx.draw_networkx_nodes(
    G, pos,
    nodelist=list(load_buses),
    node_shape="^",
    node_color="lightgreen",
    node_size=450
)

# Bus labels
labels = {
    bus: f"Bus {bus + 1}"
    for bus in G.nodes()
}

nx.draw_networkx_labels(
    G, pos,
    labels=labels,
    font_size=9,
    font_weight="bold"
)

# Legend
legend_items = [
    Line2D([0], [0], color="steelblue", lw=2,
           label="Transmission Line"),
    Line2D([0], [0], color="darkorange", lw=2,
           linestyle="--", label="Transformer"),
    Line2D([0], [0], marker="o", color="w",
           markerfacecolor="lightblue", markeredgecolor="black",
           markersize=10, label="Bus"),
    Line2D([0], [0], marker="s", color="w",
           markerfacecolor="tomato", markersize=10,
           label="Generator Bus"),
    Line2D([0], [0], marker="*", color="w",
           markerfacecolor="gold", markersize=13,
           label="External Grid"),
    Line2D([0], [0], marker="^", color="w",
           markerfacecolor="lightgreen", markersize=10,
           label="Load Bus")
]

plt.legend(handles=legend_items, loc="upper left")
plt.title(
    "IEEE 14-Bus Power System – Single-Line Network Diagram",
    fontsize=15,
    fontweight="bold"
)

plt.axis("off")
plt.tight_layout()

# Save high-resolution diagram
plt.savefig(
    "IEEE14_Single_Line_Diagram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("IEEE 14-Bus diagram generated successfully.")
```

```python

# STEP 18: Identify candidate capacitor locations

# Find load buses
load_buses = set(verified_net.load.bus)

# Exclude generator buses
generator_buses = set(verified_net.gen.bus)

# Exclude external grid buses
external_buses = set(verified_net.ext_grid.bus)

# Candidate locations
candidate_buses = sorted(
    load_buses - generator_buses - external_buses
)

print("CANDIDATE CAPACITOR LOCATIONS")

for bus in candidate_buses:
    print(f"Bus {bus + 1}")
```

```python
# STEP 19: Capacitor location sensitivity analysis

import copy

# Reference results
reference_net = copy.deepcopy(verified_net)

reference_voltage = reference_net.res_bus.vm_pu.copy()

reference_active_losses = (
    reference_net.res_line.pl_mw.sum()
    + reference_net.res_trafo.pl_mw.sum()
)

reference_reactive_losses = (
    reference_net.res_line.ql_mvar.sum()
    + reference_net.res_trafo.ql_mvar.sum()
)

sensitivity_results = []

for bus in candidate_buses:

    test_net = copy.deepcopy(reference_net)

    # Add 1 MVAr capacitive shunt
    pp.create_shunt(
        test_net,
        bus=bus,
        q_mvar=-1.0,
        p_mw=0.0,
        name=f"Test Capacitor Bus {bus + 1}"
    )

    try:
        pp.runpp(
            test_net,
            algorithm="nr",
            calculate_voltage_angles=True,
            enforce_q_lims=True
        )

        if test_net.converged:

            new_voltage = test_net.res_bus.vm_pu.at[bus]

            new_active_losses = (
                test_net.res_line.pl_mw.sum()
                + test_net.res_trafo.pl_mw.sum()
            )

            new_reactive_losses = (
                test_net.res_line.ql_mvar.sum()
                + test_net.res_trafo.ql_mvar.sum()
            )

            sensitivity_results.append({
                "Bus": bus + 1,
                "Original Voltage (p.u.)":
                    reference_voltage.at[bus],
                "New Voltage (p.u.)":
                    new_voltage,
                "Voltage Change (p.u.)":
                    new_voltage - reference_voltage.at[bus],
                "Active Loss Change (MW)":
                    new_active_losses - reference_active_losses,
                "Reactive Loss Change (MVAr)":
                    new_reactive_losses - reference_reactive_losses
            })

    except Exception as e:
        print(f"Load flow failed at Bus {bus + 1}: {e}")

sensitivity_df = pd.DataFrame(sensitivity_results)

print("CAPACITOR LOCATION SENSITIVITY RESULTS")

display(
    sensitivity_df.round(6)
)
```

```python
# STEP 20: Plot voltage improvement from 1 MVAr capacitor

plt.figure(figsize=(10, 5))

plt.bar(
    sensitivity_df["Bus"].astype(str),
    sensitivity_df["Voltage Change (p.u.)"]
)

plt.axhline(0, color="black", linewidth=0.8)

plt.xlabel("Candidate Bus")
plt.ylabel("Voltage Change (p.u.)")
plt.title("Voltage Sensitivity to 1 MVAr Capacitor Installation")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# STEP 21: Capacitor Rating Optimisation at Bus 14

import copy
import pandas as pd
import numpy as np
import pandapower as pp
import matplotlib.pyplot as plt

# Use the verified voltage-controlled network
base_net = copy.deepcopy(verified_net)

# Pandapower bus index 13 corresponds to Bus 14
target_bus = 13

# Capacitor ratings in MVAr
capacitor_ratings = range(0, 11)

rating_results = []

for rating in capacitor_ratings:

    test_net = copy.deepcopy(base_net)

    # Add capacitor bank when rating is greater than zero
    if rating > 0:
        pp.create_shunt(
            test_net,
            bus=target_bus,
            q_mvar=-float(rating),
            p_mw=0.0,
            name=f"Capacitor_{rating}_MVAr"
        )

    # Run load flow with generator Q limits enforced
    pp.runpp(
        test_net,
        algorithm="nr",
        calculate_voltage_angles=True,
        enforce_q_lims=True
    )

    if test_net.converged:

        # Active power losses
        active_losses = (
            test_net.res_line.pl_mw.sum()
            + test_net.res_trafo.pl_mw.sum()
        )

        # Reactive power losses
        reactive_losses = (
            test_net.res_line.ql_mvar.sum()
            + test_net.res_trafo.ql_mvar.sum()
        )

        # Bus voltage information
        voltages = test_net.res_bus.vm_pu

        # Generator reactive power
        generator_q = test_net.res_gen.q_mvar.sum()

        # Voltage violations
        violations = (
            (voltages < test_net.bus.min_vm_pu - 1e-6)
            | (voltages > test_net.bus.max_vm_pu + 1e-6)
        ).sum()

        rating_results.append({
            "Capacitor (MVAr)": rating,
            "Bus 14 Voltage (p.u.)": voltages.at[target_bus],
            "Minimum Voltage (p.u.)": voltages.min(),
            "Maximum Voltage (p.u.)": voltages.max(),
            "Active Losses (MW)": active_losses,
            "Reactive Losses (MVAr)": reactive_losses,
            "Generator Q Output (MVAr)": generator_q,
            "Voltage Violations": violations
        })

    else:
        print(f"Load flow did not converge at {rating} MVAr")

rating_df = pd.DataFrame(rating_results)

print("CAPACITOR RATING ANALYSIS - BUS 14")

display(rating_df.round(6))
```

```python

# STEP 25: Fine capacitor rating optimisation at Bus 14

fine_ratings = np.arange(6.0, 10.01, 0.25)

fine_results = []

for rating in fine_ratings:

    test_net = copy.deepcopy(verified_net)

    pp.create_shunt(
        test_net,
        bus=13,
        q_mvar=-float(rating),
        p_mw=0.0,
        name=f"Bus14_Capacitor_{rating:.2f}"
    )

    pp.runpp(
        test_net,
        algorithm="nr",
        calculate_voltage_angles=True,
        enforce_q_lims=True
    )

    if test_net.converged:

        v = test_net.res_bus.vm_pu

        p_loss = (
            test_net.res_line.pl_mw.sum()
            + test_net.res_trafo.pl_mw.sum()
        )

        q_loss = (
            test_net.res_line.ql_mvar.sum()
            + test_net.res_trafo.ql_mvar.sum()
        )

        fine_results.append({
            "Capacitor (MVAr)": rating,
            "Bus 14 Voltage (p.u.)": v.at[13],
            "Minimum Voltage (p.u.)": v.min(),
            "Maximum Voltage (p.u.)": v.max(),
            "Active Losses (MW)": p_loss,
            "Reactive Losses (MVAr)": q_loss
        })

fine_df = pd.DataFrame(fine_results)

display(fine_df.round(7))

# Identify minimum active loss
best_fine = fine_df.loc[
    fine_df["Active Losses (MW)"].idxmin()
]

print("\nBEST FINE-SWEEP RESULT")
display(best_fine.to_frame("Value"))
```

```python

# STEP 32: Performance under different loading conditions

import copy
import pandas as pd
import pandapower as pp

loading_levels = [0.8, 1.0, 1.2]

scenario_results = []

for factor in loading_levels:

    # Uncompensated network
    base_case = copy.deepcopy(verified_net)

    base_case.load["p_mw"] *= factor
    base_case.load["q_mvar"] *= factor

    pp.runpp(
        base_case,
        algorithm="nr",
        calculate_voltage_angles=True,
        enforce_q_lims=True
    )

    # Compensated network
    cap_case = copy.deepcopy(verified_net)

    cap_case.load["p_mw"] *= factor
    cap_case.load["q_mvar"] *= factor

    pp.create_shunt(
        cap_case,
        bus=13,
        q_mvar=-7.75,
        p_mw=0,
        name="7.75 MVAr Capacitor at Bus 14"
    )

    pp.runpp(
        cap_case,
        algorithm="nr",
        calculate_voltage_angles=True,
        enforce_q_lims=True
    )

    for label, case in [
        ("Without Compensation", base_case),
        ("With Compensation", cap_case)
    ]:

        p_loss = (
            case.res_line.pl_mw.sum()
            + case.res_trafo.pl_mw.sum()
        )

        q_loss = (
            case.res_line.ql_mvar.sum()
            + case.res_trafo.ql_mvar.sum()
        )

        voltage = case.res_bus.vm_pu

        violations = (
            (voltage < case.bus.min_vm_pu - 1e-6)
            | (voltage > case.bus.max_vm_pu + 1e-6)
        ).sum()

        scenario_results.append({
            "Loading (%)": factor * 100,
            "Configuration": label,
            "Active Losses (MW)": p_loss,
            "Reactive Losses (MVAr)": q_loss,
            "Minimum Voltage (p.u.)": voltage.min(),
            "Maximum Voltage (p.u.)": voltage.max(),
            "Voltage Violations": violations
        })

scenario_df = pd.DataFrame(scenario_results)

display(scenario_df.round(6))
```

```python

# STEP 33: Active loss comparison across loading conditions

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

for configuration in scenario_df["Configuration"].unique():

    subset = scenario_df[
        scenario_df["Configuration"] == configuration
    ]

    plt.plot(
        subset["Loading (%)"],
        subset["Active Losses (MW)"],
        marker="o",
        label=configuration
    )

plt.xlabel("Loading Level (%)")
plt.ylabel("Active Power Losses (MW)")
plt.title("Active Power Losses Under Different Loading Conditions")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

# STEP 34: Generator reactive power verification
# under different loading conditions

loading_levels = [0.8, 1.0, 1.2]

verification_results = []

for factor in loading_levels:

    test_net = copy.deepcopy(verified_net)

    test_net.load["p_mw"] *= factor
    test_net.load["q_mvar"] *= factor

    pp.create_shunt(
        test_net,
        bus=13,
        q_mvar=-7.75,
        p_mw=0,
        name="Bus 14 Capacitor"
    )

    pp.runpp(
        test_net,
        algorithm="nr",
        calculate_voltage_angles=True,
        enforce_q_lims=True
    )

    for idx, gen in test_net.gen.iterrows():

        q = test_net.res_gen.at[idx, "q_mvar"]

        within_limits = (
            gen.min_q_mvar - 1e-5
            <= q <=
            gen.max_q_mvar + 1e-5
        )

        verification_results.append({
            "Loading (%)": factor * 100,
            "Generator Bus": int(gen.bus) + 1,
            "Reactive Output (MVAr)": q,
            "Minimum Q": gen.min_q_mvar,
            "Maximum Q": gen.max_q_mvar,
            "Status": "Within limits" if within_limits
                      else "Violation"
        })

verification_df = pd.DataFrame(verification_results)

display(verification_df.round(6))

print(
    "All generator Q limits satisfied:",
    (verification_df["Status"] == "Within limits").all()
)
```
