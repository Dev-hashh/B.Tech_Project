```python

# STEP 2: Import and verify libraries

import pandapower as pp
import pandapower.networks as pn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Pandapower version:", pp.__version__)
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)

print("\nAll libraries imported successfully.")
```

```python

# STEP 3: Load the IEEE 14-bus test system

net = pn.case14()

print("IEEE 14-Bus Network Loaded Successfully")

print("\nNetwork Information:")
print("Number of buses:", len(net.bus))
print("Number of lines:", len(net.line))
print("Number of transformers:", len(net.trafo))
print("Number of generators:", len(net.gen))
print("Number of external grids:", len(net.ext_grid))
print("Number of loads:", len(net.load))
```

```python
# STEP 4: Inspect network data

print("BUS DATA")
display(net.bus)

print("LINE DATA")
display(net.line)

print("GENERATOR DATA")
display(net.gen)

print("LOAD DATA")
display(net.load)
```

```python
# STEP 5: Run baseline power-flow analysis

pp.runpp(
    net,
    algorithm="nr",
    calculate_voltage_angles=True
)

if net.converged:
    print("BASELINE LOAD FLOW CONVERGED SUCCESSFULLY")
else:
    print("BASELINE LOAD FLOW DID NOT CONVERGE")
```

```python
# STEP 6: Extract bus voltage results

bus_results = net.bus[["name"]].join(net.res_bus)

print("BASELINE BUS VOLTAGE PROFILE")

display(
    bus_results[["name", "vm_pu", "va_degree"]]
)
```

```python
# STEP 7: Calculate transmission losses

line_losses = net.res_line["pl_mw"].sum()

transformer_losses = (
    net.res_trafo["pl_mw"].sum()
    if len(net.res_trafo) > 0 else 0
)

total_losses = line_losses + transformer_losses

print("BASELINE SYSTEM LOSSES")
print("Line losses:", round(line_losses, 5), "MW")
print("Transformer losses:", round(transformer_losses, 5), "MW")
print("Total active power losses:", round(total_losses, 5), "MW")
```

```python
# STEP 8: Calculate total load demand

total_active_load = net.load["p_mw"].sum()
total_reactive_load = net.load["q_mvar"].sum()

print("TOTAL LOAD DEMAND")
print("Active power demand:", round(total_active_load, 4), "MW")
print("Reactive power demand:", round(total_reactive_load, 4), "MVAr")
```

```python

# STEP 9: Plot baseline voltage profile

plt.figure(figsize=(10, 5))

bus_numbers = np.arange(1, len(net.bus) + 1)
voltages = net.res_bus["vm_pu"].values

plt.plot(
    bus_numbers,
    voltages,
    marker="o",
    linewidth=2,
    label="Baseline voltage"
)

plt.axhline(1.0, linestyle="--", label="Reference voltage")
plt.axhline(0.95, linestyle=":", label="Lower reference limit")
plt.axhline(1.05, linestyle=":", label="Upper reference limit")

plt.xlabel("Bus Number")
plt.ylabel("Voltage Magnitude (p.u.)")
plt.title("Baseline Voltage Profile – IEEE 14-Bus System")

plt.xticks(bus_numbers)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print("Minimum voltage:", round(voltages.min(), 4), "p.u.")
print("Maximum voltage:", round(voltages.max(), 4), "p.u.")
```

```python

# STEP 10: Save baseline results

import copy

baseline_net = copy.deepcopy(net)

baseline_voltage = net.res_bus["vm_pu"].copy()
baseline_angle = net.res_bus["va_degree"].copy()

baseline_summary = pd.DataFrame({
    "Parameter": [
        "Total Active Load (MW)",
        "Total Reactive Load (MVAr)",
        "Line Losses (MW)",
        "Transformer Losses (MW)",
        "Total Active Power Losses (MW)",
        "Minimum Bus Voltage (p.u.)",
        "Maximum Bus Voltage (p.u.)"
    ],
    "Baseline Value": [
        total_active_load,
        total_reactive_load,
        line_losses,
        transformer_losses,
        total_losses,
        voltages.min(),
        voltages.max()
    ]
})

display(baseline_summary)

baseline_summary.to_csv(
    "IEEE14_Baseline_Summary.csv",
    index=False
)

bus_results.to_csv(
    "IEEE14_Baseline_Bus_Results.csv"
)

print("Baseline results saved successfully.")
```

```python

# STEP 11: Generate IEEE 14-Bus Single-Line Diagram

import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Create network graph
G = nx.MultiGraph()

# Add buses
for bus in net.bus.index:
    G.add_node(bus)

# Extract transmission line connections
line_edges = [
    (row.from_bus, row.to_bus)
    for _, row in net.line.iterrows()
]

# Extract transformer connections
trafo_edges = [
    (row.hv_bus, row.lv_bus)
    for _, row in net.trafo.iterrows()
]

G.add_edges_from(line_edges)
G.add_edges_from(trafo_edges)

# Generate automatic network layout
pos = nx.spring_layout(G, seed=14, k=1.3, iterations=200)

# Identify generator, external grid and load buses
gen_buses = set(net.gen.bus)
ext_buses = set(net.ext_grid.bus)
load_buses = set(net.load.bus)

# Draw diagram
plt.figure(figsize=(13, 9))

# Transmission lines
nx.draw_networkx_edges(
    G, pos,
    edgelist=line_edges,
    edge_color="steelblue",
    width=2
)

# Transformers
nx.draw_networkx_edges(
    G, pos,
    edgelist=trafo_edges,
    edge_color="darkorange",
    width=2.5,
    style="dashed"
)

# All buses
nx.draw_networkx_nodes(
    G, pos,
    node_size=650,
    node_color="lightblue",
    edgecolors="black"
)

# Generator buses
nx.draw_networkx_nodes(
    G, pos,
    nodelist=list(gen_buses),
    node_shape="s",
    node_color="tomato",
    node_size=700
)

# External grid bus
nx.draw_networkx_nodes(
    G, pos,
    nodelist=list(ext_buses),
    node_shape="*",
    node_color="gold",
    node_size=1100
)

# Load buses
nx.draw_networkx_nodes(
    G, pos,
    nodelist=list(load_buses),
    node_shape="^",
    node_color="lightgreen",
    node_size=450
)

# Bus labels
labels = {
    bus: f"Bus {bus + 1}"
    for bus in G.nodes()
}

nx.draw_networkx_labels(
    G, pos,
    labels=labels,
    font_size=9,
    font_weight="bold"
)

# Legend
legend_items = [
    Line2D([0], [0], color="steelblue", lw=2,
           label="Transmission Line"),
    Line2D([0], [0], color="darkorange", lw=2,
           linestyle="--", label="Transformer"),
    Line2D([0], [0], marker="o", color="w",
           markerfacecolor="lightblue", markeredgecolor="black",
           markersize=10, label="Bus"),
    Line2D([0], [0], marker="s", color="w",
           markerfacecolor="tomato", markersize=10,
           label="Generator Bus"),
    Line2D([0], [0], marker="*", color="w",
           markerfacecolor="gold", markersize=13,
           label="External Grid"),
    Line2D([0], [0], marker="^", color="w",
           markerfacecolor="lightgreen", markersize=10,
           label="Load Bus")
]

plt.legend(handles=legend_items, loc="upper left")
plt.title(
    "IEEE 14-Bus Power System – Single-Line Network Diagram",
    fontsize=15,
    fontweight="bold"
)

plt.axis("off")
plt.tight_layout()

# Save high-resolution diagram
plt.savefig(
    "IEEE14_Single_Line_Diagram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("IEEE 14-Bus diagram generated successfully.")
```

```python

# STEP 12: Investigate voltage limits and generator operation

print("BUS VOLTAGE LIMITS")

voltage_check = net.bus[
    ["name", "min_vm_pu", "max_vm_pu"]
].join(net.res_bus[["vm_pu"]])

voltage_check["status"] = np.where(
    voltage_check["vm_pu"] >
    voltage_check["max_vm_pu"],
    "Above maximum",
    np.where(
        voltage_check["vm_pu"] <
        voltage_check["min_vm_pu"],
        "Below minimum",
        "Within limits"
    )
)

display(voltage_check)

print("\nGENERATOR PARAMETERS")
display(net.gen)

print("\nGENERATOR OPERATING RESULTS")
display(net.res_gen)

print("\nEXTERNAL GRID RESULTS")
display(net.res_ext_grid)
```

```python
# STEP 13: Inspect individual transformer losses

trafo_results = net.trafo[
    ["hv_bus", "lv_bus"]
].join(net.res_trafo)

display(
    trafo_results[
        ["hv_bus", "lv_bus", "p_hv_mw",
         "p_lv_mw", "pl_mw", "ql_mvar"]
    ]
)
```

```python
# STEP 14: Create a voltage-controlled test scenario

import copy

controlled_net = copy.deepcopy(baseline_net)

# Adjust generator voltage setpoints
# Pandapower uses zero-based bus indices internally.

controlled_net.gen.loc[
    controlled_net.gen.bus == 5, "vm_pu"
] = 1.05

controlled_net.gen.loc[
    controlled_net.gen.bus == 7, "vm_pu"
] = 1.05

# Run load flow
pp.runpp(
    controlled_net,
    algorithm="nr",
    calculate_voltage_angles=True
)

print("Voltage-controlled scenario converged:",
      controlled_net.converged)

# Display new voltage profile
controlled_voltage = controlled_net.res_bus["vm_pu"]

comparison = pd.DataFrame({
    "Bus": np.arange(1, 15),
    "Original Voltage (p.u.)":
        baseline_voltage.values,
    "Controlled Voltage (p.u.)":
        controlled_voltage.values
})

display(comparison)
```

```python
# STEP 15: Compare original and controlled voltage profiles

plt.figure(figsize=(11, 5))

buses = np.arange(1, 15)

plt.plot(
    buses,
    baseline_voltage.values,
    marker="o",
    label="Original IEEE 14-bus"
)

plt.plot(
    buses,
    controlled_voltage.values,
    marker="s",
    label="Voltage-controlled scenario"
)

plt.axhline(1.06, linestyle="--", label="Maximum voltage")
plt.axhline(0.94, linestyle="--", label="Minimum voltage")

plt.xlabel("Bus Number")
plt.ylabel("Voltage Magnitude (p.u.)")
plt.title("Voltage Profile Comparison")
plt.xticks(buses)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

# STEP 16: Check generator reactive power limits

# Create a fresh controlled scenario
verified_net = copy.deepcopy(baseline_net)

verified_net.gen.loc[
    verified_net.gen.bus == 5, "vm_pu"
] = 1.05

verified_net.gen.loc[
    verified_net.gen.bus == 7, "vm_pu"
] = 1.05

# Enforce generator reactive power limits
pp.runpp(
    verified_net,
    algorithm="nr",
    calculate_voltage_angles=True,
    enforce_q_lims=True
)

print("Load flow converged:", verified_net.converged)

# Generator reactive power comparison
generator_check = verified_net.gen[
    ["bus", "min_q_mvar", "max_q_mvar"]
].join(verified_net.res_gen[["q_mvar"]])

generator_check["Q limit status"] = np.where(
    generator_check["q_mvar"] >
    generator_check["max_q_mvar"] + 1e-5,
    "Above maximum",
    np.where(
        generator_check["q_mvar"] <
        generator_check["min_q_mvar"] - 1e-5,
        "Below minimum",
        "Within limits"
    )
)

display(generator_check)

# Check voltage limits again
voltage_check = verified_net.bus[
    ["min_vm_pu", "max_vm_pu"]
].join(verified_net.res_bus[["vm_pu"]])

display(voltage_check)
```

```python

# STEP 17: Accurate active and reactive loss calculation

line_active_loss = verified_net.res_line.pl_mw.sum()
trafo_active_loss = verified_net.res_trafo.pl_mw.sum()

line_reactive_loss = verified_net.res_line.ql_mvar.sum()
trafo_reactive_loss = verified_net.res_trafo.ql_mvar.sum()

print("ACTIVE POWER LOSSES")
print("Line losses:", line_active_loss, "MW")
print("Transformer losses:", trafo_active_loss, "MW")
print("Total active losses:",
      line_active_loss + trafo_active_loss, "MW")

print("\nREACTIVE POWER LOSSES")
print("Line reactive losses:", line_reactive_loss, "MVAr")
print("Transformer reactive losses:", trafo_reactive_loss, "MVAr")
print("Total reactive losses:",
      line_reactive_loss + trafo_reactive_loss, "MVAr")
```

```python

# STEP 18: Identify candidate capacitor locations

# Find load buses
load_buses = set(verified_net.load.bus)

# Exclude generator buses
generator_buses = set(verified_net.gen.bus)

# Exclude external grid buses
external_buses = set(verified_net.ext_grid.bus)

# Candidate locations
candidate_buses = sorted(
    load_buses - generator_buses - external_buses
)

print("CANDIDATE CAPACITOR LOCATIONS")

for bus in candidate_buses:
    print(f"Bus {bus + 1}")
```

```python
# STEP 19: Capacitor location sensitivity analysis

import copy

# Reference results
reference_net = copy.deepcopy(verified_net)

reference_voltage = reference_net.res_bus.vm_pu.copy()

reference_active_losses = (
    reference_net.res_line.pl_mw.sum()
    + reference_net.res_trafo.pl_mw.sum()
)

reference_reactive_losses = (
    reference_net.res_line.ql_mvar.sum()
    + reference_net.res_trafo.ql_mvar.sum()
)

sensitivity_results = []

for bus in candidate_buses:

    test_net = copy.deepcopy(reference_net)

    # Add 1 MVAr capacitive shunt
    pp.create_shunt(
        test_net,
        bus=bus,
        q_mvar=-1.0,
        p_mw=0.0,
        name=f"Test Capacitor Bus {bus + 1}"
    )

    try:
        pp.runpp(
            test_net,
            algorithm="nr",
            calculate_voltage_angles=True,
            enforce_q_lims=True
        )

        if test_net.converged:

            new_voltage = test_net.res_bus.vm_pu.at[bus]

            new_active_losses = (
                test_net.res_line.pl_mw.sum()
                + test_net.res_trafo.pl_mw.sum()
            )

            new_reactive_losses = (
                test_net.res_line.ql_mvar.sum()
                + test_net.res_trafo.ql_mvar.sum()
            )

            sensitivity_results.append({
                "Bus": bus + 1,
                "Original Voltage (p.u.)":
                    reference_voltage.at[bus],
                "New Voltage (p.u.)":
                    new_voltage,
                "Voltage Change (p.u.)":
                    new_voltage - reference_voltage.at[bus],
                "Active Loss Change (MW)":
                    new_active_losses - reference_active_losses,
                "Reactive Loss Change (MVAr)":
                    new_reactive_losses - reference_reactive_losses
            })

    except Exception as e:
        print(f"Load flow failed at Bus {bus + 1}: {e}")

sensitivity_df = pd.DataFrame(sensitivity_results)

print("CAPACITOR LOCATION SENSITIVITY RESULTS")

display(
    sensitivity_df.round(6)
)
```

```python
# STEP 20: Plot voltage improvement from 1 MVAr capacitor

plt.figure(figsize=(10, 5))

plt.bar(
    sensitivity_df["Bus"].astype(str),
    sensitivity_df["Voltage Change (p.u.)"]
)

plt.axhline(0, color="black", linewidth=0.8)

plt.xlabel("Candidate Bus")
plt.ylabel("Voltage Change (p.u.)")
plt.title("Voltage Sensitivity to 1 MVAr Capacitor Installation")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
